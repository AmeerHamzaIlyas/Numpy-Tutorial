```python
import numpy as np
```

```python
python_list = [10,20,30,40,50]
numpy_array  = np.array([10,20,30,40,50])
print("Python List",python_list)
print("Numpy array",numpy_array)
print("Type of python list",type(python_list))
print("Type of numpy array",type(numpy_array))
print("python lists (require loop):",[x + 5 for x in python_list])
print("numpy arrays (vectorized):",numpy_array + 5)

```

```python
import time
size = 1000000
python_list= list(range(size))
numpy_array = np.array(size)
print("----------Performance Comparision------------")
print(f"size of array {size:,} elements\n")

start_time = time.time()
result_python = [x + 1 for x in python_list]
time_python = time.time()- start_time
print(f"python array list addition time is {time_python:6f} seconds")


tart_time = time.time()
result_numpy = numpy_array + 1
time_numpy = time.time()- start_time
print(f"Numpy array  addition time is {time_numpy:6f} seconds")

print(f"Numpy is {time_python/time_numpy:.1f}X faster")
```

Numpy Arrays

```python
#1
list_like_array = np.array([1,2,3,4])
print("like list array",list_like_array)

#2
arange = np.arange(1,10,2) # start,stop, step
print("like (range of) array",arange)

#3
linespace = np.linspace(1,12,4) # start, stop, no. of elements
print("using linespace",linespace)

#4
zeros = np.zeros(5) # No. of zeros
print("using zeros",zeros)

#5
ones = np.ones(5) # No. of ones
print("using ones",ones)

#6
full = np.full(5,7) # shape numbers, full values
print("using full (all 7s)",full)

#7
identity = np.eye(5) 
print("identity matrix\n",identity)

#8
random_array = np.random.rand(7) #
print("Generating random number array",random_array)

#9
random_int_array = np.random.randint(1,10,size=5) # low, high, size (by default is one when we not mension)
print("Generating random integer number array",random_int_array)
```

Numpy Data Types

```python
arr_int = np.array([1,2,3])
arr_float = np.array([1.2,2.6,3.9])
arr_mixed = np.array([1,2.5,3]) # converted to float

print(arr_int.dtype)
print(arr_float.dtype)
print(arr_mixed.dtype)

#speccifing data types
new_arr1= arr_float.astype(np.float32)
new_arr2 = arr_float.astype(np.int64)

print(new_arr1,new_arr1.dtype)
print(new_arr2,new_arr2.dtype)

print("int32/float32 = takes 4 bytes , int64/float64 = takes 8 bytes")



```
